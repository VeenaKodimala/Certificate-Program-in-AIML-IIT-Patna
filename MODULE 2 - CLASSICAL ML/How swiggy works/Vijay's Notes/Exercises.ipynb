Module 1: Concept 1 — Creating a NumPy Array

```python
import numpy as np

# A small shop's daily sales (INR)
sales = np.array([1200, 850, 1500, 970, 2100, 1800, 1350])

print(sales)          # [1200  850 1500  970 2100 1800 1350]
print(type(sales))    # <class 'numpy.ndarray'>
print(sales.dtype)    # int64
print(sales.shape)    # (7,)  — 7 elements, 1 dimension
```

Concept 2 — Aggregations (sum, mean, percentile)

```python
total_sales  = np.sum(sales)           # 9770
avg_sales    = np.mean(sales)          # 1395.71
median_sales = np.median(sales)        # 1350
best_day     = np.max(sales)           # 2100
worst_day    = np.min(sales)           # 850
top10pct     = np.percentile(sales, 90) # 1980  (90% of days sold less than this)
```

```python
print(total_sales)
print(avg_sales)
print(median_sales)
print(best_day)
print(worst_day)
print(top10pct)
```

Concept 3 — Boolean Masks (Filter Without a Loop)

```python
# Which days had sales above 1400?
mask = sales > 1400
print(mask)           # [False False  True False  True  True False]

# Apply mask — returns only the matching values
high_sales = sales[mask]
print(high_sales)     # [1500 2100 1800]

# One-liner version (most common in practice)
sales[sales > 1400]   # [1500 2100 1800]
```

Concept 4 — Vectorised Operations (No Loops Needed)

```python
sales_with_gst = sales * 1.18
print(sales_with_gst)
# [1416.  1003.  1770.  1144.6  2478.  2124.  1593. ]

# Combine two arrays element-wise
costs = np.array([600, 400, 700, 500, 900, 750, 600])
profit = sales - costs
print(profit)  # [600 450 800 470 1200 1050 750]
```

Concept 5 — Speed Demo (Loop vs Vectorised)

```python
import time

big = np.random.randint(100, 1000, 1_000_000)  # 1 million numbers

# Python loop
start = time.time()
total = 0
for x in big:
    total += x
print(f'Loop:  {time.time()-start:.4f}s')

# NumPy vectorised
start = time.time()
total = np.sum(big)
print(f'NumPy: {time.time()-start:.4f}s')
```

Concept 1 — Loading a CSV Correctly

```python
import pandas as pd

# Create a small example CSV in memory
from io import StringIO
data = '''order_id,item,price,order_date,rating
1,Coffee,120,2024-01-15,4.5
2,Sandwich,250,2024-01-15,4.0
3,Juice,90,2024-01-16,
4,Coffee,120,2024-01-17,5.0
5,Cake,180,2024-01-17,3.5'''

# Without parse_dates — order_date loads as object (string)
df = pd.read_csv(StringIO(data))
print(df['order_date'].dtype)  # object  <-- BAD

# With parse_dates — order_date loads as datetime64
df = pd.read_csv(StringIO(data), parse_dates=['order_date'])
print(df['order_date'].dtype)  # datetime64[ns]  <-- GOOD
```

Module 3\
Concept 1 — Single Condition Boolean Filter

```python
import pandas as pd

df = pd.DataFrame({
    'name':  ['Priya','Rahul','Anjali','Vikram','Sneha'],
    'score': [88, 72, 95, 61, 84],
    'city':  ['Mumbai','Delhi','Mumbai','Pune','Delhi'],
    'passed': [True, True, True, False, True]
})

# Filter: students who scored above 80
high_scorers = df[df['score'] > 80]
print(high_scorers)
```

Concept 2 — Multi-Condition Filters (AND, OR)

```python
# AND condition: Mumbai students who scored above 80
result = df[(df['city'] == 'Mumbai') & (df['score'] > 80)]
# Priya (Mumbai, 88), Anjali (Mumbai, 95)

# OR condition: Mumbai OR score above 90
result = df[(df['city'] == 'Mumbai') | (df['score'] > 90)]
# Priya, Anjali, Sneha (Delhi, 84 — NO, this one fails both!)

# WRONG — missing parentheses — gives wrong result silently!
# df[df['city'] == 'Mumbai' & df['score'] > 80]   # BAD
```

Concept 3 — .loc vs .iloc

```python
# .loc — select by label
df.loc[0:2, ['name', 'score']]   # rows 0,1,2 (inclusive!) — name and score columns

# .iloc — select by position
df.iloc[0:2, 0:2]   # rows 0,1 (exclusive end!) — first 2 columns

# .loc with condition — most common pattern
df.loc[df['score'] > 80, ['name', 'city']]
```

Concept 4 — sort_values()


```python
# Sort by score, highest first
df.sort_values('score', ascending=False)

# Multi-key sort: city alphabetically, then score descending within city
df.sort_values(['city', 'score'], ascending=[True, False])

# Get top 3 students
df.sort_values('score', ascending=False).head(3)
```

Concept 5 — Chaining Filters, Sort, and Selection

```python
result = (
    df[df['city'] == 'Mumbai']    # Step 1: Filter city
    .sort_values('score',          # Step 2: Sort by score
                 ascending=False)
    [['name', 'score']]            # Step 3: Keep only name & score
    .head(3)                       # Step 4: Top 3 only
)
```

Module 4\
Concept 1 — Basic groupby + Single Aggregation

```python
import pandas as pd

df = pd.DataFrame({
    'name':    ['Priya','Rahul','Anjali','Vikram','Sneha','Arjun'],
    'score':   [88, 72, 95, 61, 84, 79],
    'city':    ['Mumbai','Delhi','Mumbai','Pune','Delhi','Pune'],
    'subject': ['Math','Science','Math','Science','Math','Science'],
    'fees':    [15000, 12000, 18000, 10000, 13000, 11000]
})

# Average score by city
df.groupby('city')['score'].mean()
```

Concept 2 — Multiple Aggregations with .agg()

```python
# For each city: total fees, average score, number of students
result = df.groupby('city').agg(
    total_fees   = ('fees',  'sum'),
    avg_score    = ('score', 'mean'),
    student_count= ('name',  'count')
)
print(result)
```

Concept 3 — Multi-Column GroupBy

```python
# Average score by city AND subject
df.groupby(['city', 'subject'])['score'].mean()

# city    subject
# Delhi   Math       84.0
#         Science    72.0
# Mumbai  Math       91.5
# Pune    Science    70.0

# Reset index to get a flat DataFrame
result = df.groupby(['city','subject'])['score'].mean().reset_index()
```

Concept 4 — Pivot Tables

```python
pivot = pd.pivot_table(
    df,
    values  = 'fees',         # What to aggregate
    index   = 'city',         # Rows
    columns = 'subject',      # Columns
    aggfunc = 'sum',          # How to aggregate
    fill_value = 0            # Replace NaN with 0
)
```

Concept 5 — transform() for % Contribution

```python
# Step 1: Get city total for EVERY row using transform
df['city_total_fees'] = df.groupby('city')['fees'].transform('sum')

# Step 2: Calculate percentage
df['fee_share'] = (df['fees'] / df['city_total_fees'] * 100).round(1)

print(df[['name', 'city', 'fees', 'city_total_fees', 'fee_share']])
```

Module 5\
Concept 1 — Detecting Nulls

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'product':  ['Burger','Pizza','Dosa','Biryani','Wrap'],
    'price':    [180, 299, None, 350, 149],
    'rating':   [4.2, None, 4.5, 4.8, None],
    'stock':    [50, 20, 80, None, 30],
    'category': ['Burger','Pizza','South Indian',None,'Rolls']
})

# Count nulls per column
print(df.isnull().sum())
print((df.isnull().sum() / len(df) * 100).round(1))
```

Concept 2 — fillna() Strategies

```python
# Fill price with median (robust to outliers)
df['price'] = df['price'].fillna(df['price'].median())

# Fill rating with mean
df['rating'] = df['rating'].fillna(df['rating'].mean().round(1))

# Fill category with mode
df['category'] = df['category'].fillna(df['category'].mode()[0])

# Verify — should now be 0
print(df.isnull().sum())
```

Concept 3 — Detecting and Removing Duplicates

```python
orders = pd.DataFrame({
    'order_id': [101, 102, 103, 101, 104],   # 101 appears twice!
    'item':     ['Burger','Pizza','Dosa','Burger','Wrap'],
    'amount':   [180, 299, 90, 180, 149]
})
# Detect duplicates
print(orders.duplicated(subset='order_id').sum())  # 1 duplicate

# See which rows are duplicated
print(orders[orders.duplicated(subset='order_id', keep=False)])

# Remove — keep first occurrence
clean = orders.drop_duplicates(subset='order_id', keep='first')
print(f'Before: {len(orders)} rows. After: {len(clean)} rows.')
# Before: 5 rows. After: 4 rows.
```

Concept 4 — Fixing Impossible / Invalid Values

```python
# Inject bad data
df.loc[0, 'rating'] = 6.5   # Impossible — max is 5
df.loc[1, 'price']  = -50   # Impossible — negative price

# Detect
print(df[df['rating'] > 5])       # Find rating outliers
print(df[df['price'] < 0])        # Find negative prices

# Fix: replace with NaN, then fill
df.loc[df['rating'] > 5, 'rating'] = np.nan
df.loc[df['price'] < 0, 'price']   = np.nan

# Now fill the newly created nulls
df['rating'] = df['rating'].fillna(df['rating'].median())
df['price']  = df['price'].fillna(df['price'].median())
```

Concept 5 — Outlier Detection (3-Sigma Rule)

```python
prices = np.array([180, 299, 90, 350, 149, 180, 5000, 250, 200, 175])

mean = prices.mean()   # 487.3  <-- inflated by the 5000
std  = prices.std()    # 1433.6

# 3-sigma bounds
lower = mean - 3 * std   # -3714  (negative, so lower bound is irrelevant here)
upper = mean + 3 * std   #  4788

# Find outliers
outliers = prices[prices > upper]
print(outliers)   # [5000]

# Mean WITHOUT the outlier
clean = prices[prices <= upper]
print(f'With outlier: mean={mean:.0f}  Without: mean={clean.mean():.0f}')
```
