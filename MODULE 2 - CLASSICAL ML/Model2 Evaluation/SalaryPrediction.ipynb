```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay,classification_report
```

```python
data=[{'age':25, 'annual_income_lakhs':3.5, 'credit_score':620, 'employment_type':'Salaried', 'loan_approved':0},
 {'age':35, 'annual_income_lakhs':7.0, 'credit_score':750, 'employment_type':'Salaried', 'loan_approved':1},
 {'age':28, 'annual_income_lakhs':4.5, 'credit_score':680, 'employment_type':'Self-Employed', 'loan_approved':0},
 {'age':45, 'annual_income_lakhs':12.0, 'credit_score':810, 'employment_type':'Salaried', 'loan_approved':1},
 {'age':30, 'annual_income_lakhs':5.5, 'credit_score':710, 'employment_type':'Salaried', 'loan_approved':1},
 {'age':22, 'annual_income_lakhs':2.8, 'credit_score':590, 'employment_type':'Self-Employed', 'loan_approved':0},
 {'age':40, 'annual_income_lakhs':9.0, 'credit_score':770, 'employment_type':'Salaried', 'loan_approved':1},
 {'age':33, 'annual_income_lakhs':6.0, 'credit_score':640, 'employment_type':'Self-Employed', 'loan_approved':0}
 ]

df = pd.DataFrame(data)
df.head()
```

```python
#Label-encoding
df['employment_type'] = df['employment_type'].map({'Salaried':1, 'Self-Employed':0})
```

```python
df.head()
```

```python
x=df.drop('loan_approved', axis=1)
y=df['loan_approved']
```

```python
display(x)
display(y)
```

```python
xTrain,xTest,yTrain,yTest = train_test_split(x,y,random_state=42,test_size=0.25)
```

```python
from matplotlib import pyplot as py
dscTree = DecisionTreeClassifier(max_depth=3, random_state=42)
dscTree.fit(xTrain,yTrain)

yTest_Pred = dscTree.predict(xTest)

print(yTest)
print(yTest_Pred)

confusionMatrix = confusion_matrix(yTest,yTest_Pred)
disp = ConfusionMatrixDisplay(confusionMatrix)

disp.plot()
py.show()

print(classification_report(yTest,yTest_Pred, target_names=['loan approved','loan not-approved']))


```
