# Bagging (Bootstrap Aggregating)

## Definition

**Bagging (Bootstrap Aggregating)** is an ensemble learning technique in which multiple models are trained on different random subsets of the training data, and their predictions are combined to produce a more accurate and stable final prediction.

> **One-line definition:**  
> Train multiple models independently on different random samples of the dataset and combine their predictions.

---

# Why Do We Need Bagging?

A single machine learning model (especially a Decision Tree) can easily **overfit** the training data.

Bagging helps by:

- Reducing overfitting
- Reducing variance
- Improving prediction accuracy
- Making the model more stable

Instead of relying on one model, we rely on the collective opinion of many models.

---

# How Bagging Works

Suppose we have a dataset containing **1000 samples**.

Instead of training one Decision Tree:

```
1000 Samples
      │
      ▼
Decision Tree
      │
Prediction
```

Bagging creates multiple random datasets from the original dataset.

```
Original Dataset
        │
        ▼
───────────────────────────────
Bootstrap Sample 1 → Tree 1
Bootstrap Sample 2 → Tree 2
Bootstrap Sample 3 → Tree 3
Bootstrap Sample 4 → Tree 4
Bootstrap Sample 5 → Tree 5
───────────────────────────────
        │
        ▼
Combine Predictions
        │
        ▼
Final Prediction
```

---

# What is Bootstrap Sampling?

Bootstrap sampling means **sampling with replacement**.

### Original Dataset

```
A  B  C  D  E
```

A possible bootstrap sample is

```
A  C  C  D  E
```

Notice that:

- `C` appears twice.
- `B` is missing.

This is completely valid because sampling is done **with replacement**.

Each bootstrap sample has the **same size** as the original dataset but contains different combinations of samples.

---

# Prediction Process

Each model makes its own prediction.

Example:

```
Tree 1 → Dog
Tree 2 → Dog
Tree 3 → Cat
Tree 4 → Dog
Tree 5 → Cat
```

For **classification**, the final prediction is obtained using **Majority Voting**.

```
Dog → 3 votes
Cat → 2 votes

Final Prediction = Dog
```

For **regression**, the predictions are averaged.

Example:

```
Tree 1 → 15
Tree 2 → 18
Tree 3 → 17
Tree 4 → 16

Final Prediction

= (15 + 18 + 17 + 16) / 4
= 16.5
```

---

# Why Does Bagging Work?

Each model sees a slightly different dataset.

Therefore,

- Every model makes different mistakes.
- The mistakes tend to cancel each other.
- The final prediction becomes more reliable.

Instead of trusting one model, we trust the consensus of many models.

---

# Bagging Workflow

```
Original Dataset
        │
        ▼
Create Bootstrap Samples
        │
        ▼
Train Multiple Models
        │
        ▼
Each Model Predicts
        │
        ▼
Combine Predictions
        │
        ▼
Final Prediction
```

---

# Advantages

- Reduces overfitting
- Reduces variance
- Produces more stable predictions
- Improves accuracy
- Works very well with Decision Trees
- Models can be trained in parallel

---

# Disadvantages

- Training multiple models requires more memory.
- Prediction can be slower because many models are evaluated.
- Less interpretable than a single Decision Tree.

---

# Bagging vs Single Decision Tree

| Single Decision Tree | Bagging |
|----------------------|----------|
| One model | Many models |
| High variance | Low variance |
| Can overfit | Less prone to overfitting |
| Less stable | More stable |
| Faster | Slightly slower |

---

# Bagging vs Boosting

| Bagging | Boosting |
|---------|----------|
| Models are trained independently | Models are trained sequentially |
| Uses random bootstrap samples | Focuses on correcting previous mistakes |
| Reduces variance | Reduces bias |
| Models have equal importance | Later models have greater influence |
| Easy to parallelize | Difficult to parallelize |

---

# Real-World Example

Suppose **five doctors** independently diagnose a patient.

```
Doctor 1 → Flu
Doctor 2 → Flu
Doctor 3 → Cold
Doctor 4 → Flu
Doctor 5 → Cold
```

Majority opinion:

```
Flu
```

This is exactly how **Bagging** works.

Instead of trusting one expert, it trusts the majority of multiple experts.

---

# Key Points to Remember

- **Bagging = Bootstrap + Aggregating**
- Bootstrap means **sampling with replacement**.
- Multiple models are trained independently.
- Every model receives a different random subset of the training data.
- Classification uses **Majority Voting**.
- Regression uses **Averaging**.
- Bagging mainly reduces **variance**.
- It is commonly used with **Decision Trees**.
