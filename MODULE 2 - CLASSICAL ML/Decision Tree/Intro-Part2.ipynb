


### How Decision trees are different from Logistic Regression:
- Logistic Regression makes a prediction by combining all features into a single weighted score, whereas a Decision Tree makes a prediction by following a sequence of feature-based decision rules (questions).
- Logistic Regression: "How much does each feature contribute to the decision?"
- Decision Tree: "Which question should I ask next to separate the data?"

- If the depth of the tree is greater, then the model will be an overfitting one. If the depth is intuitive, then the model will turn out to be best model.
