Problem Statement

A team is prototyping a small semantic search feature over five short internal-policy document chunks. Each chunk has already been converted into a 4-dimensional embedding vector (a simplified stand-in for a real embedding model's output), so no embedding API call is required for this task — you will work directly with the numeric vectors provided below.

Dataset
Query embedding (for the search text "what is the process to cancel a subscription"):

query = [1.00, 0.00, 0.00, 0.00]

Document chunk embeddings:

chunk_id	chunk_text (short label)	embedding vector
chunk_1	Subscription cancellation steps	[0.90, 0.10, 0.30, 0.20]
chunk_2	Refund timeline after cancellation	[0.60, 0.50, 0.40, 0.30]
chunk_3	Office Wi-Fi setup guide	[0.20, 0.60, 0.60, 0.50]
chunk_4	How to cancel a paid plan	[0.75, 0.30, 0.30, 0.20]
chunk_5	Parking allotment policy	[0.05, 0.50, 0.50, 0.50]
Tasks
Implement a cosine_similarity(v1, v2) function from first principles (dot product divided by the product of the two vectors' norms). Do not use a prebuilt similarity function from a library (e.g., sklearn.metrics.pairwise.cosine_similarity, scipy.spatial.distance.cosine); you may use Python's math module for the square root.
Compute the similarity between the query embedding and every chunk embedding.
Write a semantic_search(query_vec, chunks, top_k) function that returns the top_k chunks ranked by similarity, most similar first.
For every chunk (not just the top-k), classify its similarity score into one of three bands: "strong" (> 0.8), "related" (0.5-0.8), or "unrelated" (< 0.5).
Expected Output
Running semantic_search(query, chunks, top_k=3) should return the 3 most similar chunks in descending similarity order, each tagged with its numeric score (rounded to 3 decimal places) and its band classification.

Constraints
Do not hardcode the similarity scores or the final ranking — compute them programmatically from the vectors given.
top_k must be a parameter, not a hardcoded 3 inside the ranking logic.

```python
import numpy as np
```

```python
userQry="what is the process to cancel a subscription"
userQryEmbedding = [1.00, 0.00, 0.00, 0.00]

embeddingDataset = {
        "Subscription cancellation steps":[0.90, 0.10, 0.30, 0.20],
    "Refund timeline after cancellation":[0.60, 0.50, 0.40, 0.30],
    "Office Wi-Fi setup guide":[0.20, 0.60, 0.60, 0.50],
    "How to cancel a paid plan":[0.75, 0.30, 0.30, 0.20],
    "Parking allotment policy":[0.05, 0.50, 0.50, 0.50]}
```

```python
def getCosineSimilarity(vec1,vec2):
    vec1 = np.array(vec1)
    vec2 = np.array(vec2)

    cosineSimilarity = np.dot(vec1,vec2)/(np.linalg.norm(vec1) * np.linalg.norm(vec1))
    return cosineSimilarity 
```

```python
keys = list(embeddingDataset.keys())
scores=[]
for i in range(len(keys)):
    emb1 = userQryEmbedding
    emb2 = embeddingDataset[keys[i]]

    cosineSimilarity = getCosineSimilarity(emb1,emb2)

    print(f"Similarity between '{userQry}' and '{keys[i]}':: {cosineSimilarity}")
```

```python
def semantic_search(query_vec, chunks, top_k=3):
    print(f"length of User query embedding:{len(userQryEmbedding)}")

    keys = list(embeddingDataset.keys())
    scores=[]

    for i in range(len(keys)):
        emb1 = userQryEmbedding
        emb2 = embeddingDataset[keys[i]]
        cosineSimilarity = getCosineSimilarity(emb1,emb2)
        scores.append((keys[i],cosineSimilarity))
    scores.sort(reverse=True)

    print(scores)

    for rank, (text,score) in enumerate(scores[:top_k],1):
        print(f"score: {score}")
        if score > 0.8:
            band = "strong"
        elif score >= 0.5 and score <= 0.8:
            band = "related"
        else:
            band = "unrelated"

        print(f"Band: {band}")            
        print(f"{rank} - {text} ::: {score}:: band: {band}")
    print()     

    
```

```python
semantic_search(userQryEmbedding,embeddingDataset,top_k=3)
```
